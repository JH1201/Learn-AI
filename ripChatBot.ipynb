```python
import re
from getpass import getpass

import requests

# 실행 시 나타나는 입력창에 입력
NOTION_TOKEN = getpass("노션 API 토큰: ").strip()
page_url = input("읽을 노션 페이지 URL: ").strip()

# 페이지 URL에서 페이지 ID 추출
last_part = page_url.split("?")[0].rstrip("/").split("/")[-1]
match = re.search(r"([0-9a-fA-F]{32})$", last_part.replace("-", ""))

if not match:
    raise ValueError("페이지 ID를 찾을 수 없습니다. 노션 페이지 링크를 확인해주세요.")

PAGE_ID = match.group(1)

headers = {
    "Authorization": f"Bearer {NOTION_TOKEN}",
    "Notion-Version": "2026-03-11",
}

# 첫 번째 블록만 요청해서 접근 가능 여부 확인
response = requests.get(
    f"https://api.notion.com/v1/blocks/{PAGE_ID}/children",
    headers=headers,
    params={"page_size": 1},
    timeout=30,
)

print("응답 상태:", response.status_code)

if response.status_code == 200:
    print("성공! 이 노션 페이지의 내용을 읽을 수 있습니다.")
else:
    print(response.json().get("message", "요청에 실패했습니다."))
```
