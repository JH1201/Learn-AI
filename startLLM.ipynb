```python
# LangChain 기초와 LCEL

import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage

# .env 파일에 저장된 환경변수를 시스템 변수로 로드
load_dotenv()

# openAI 대신 openRouter 사용
ai_key = os.getenv("OPENROUTER_API_KEY")

llm = ChatOpenAI(
    model='openrouter/free',  # 무료 모델 자동 선택
    api_key=ai_key,
    base_url="https://openrouter.ai/api/v1"
)

msg = [
    SystemMessage(content="너는 Python 강사지만 반말만 사용하는 예의없는 강사야"),
    HumanMessage(content="Python이 무엇인가요?")
]


res = llm.invoke(msg)

print(res.content)
```

```python
# prompt template / LCEL

import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# 각 처리 단계를 파이프 기호로 연결

load_dotenv()

prompt = ChatPromptTemplate.from_messages([
    ("system", "너는 핵심만 설명하는 AI 강사고 존댓말을 사용해야 돼. 설명은 1문장을 넘지 않아."),
    ("human", "{topic}을 초보자 수준으로 설명해줘")
])

llm = ChatOpenAI(
    model="openrouter/free",
    base_url="https://openrouter.ai/api/v1",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    temperature = 1.0
)

parser = StrOutputParser()

chain = prompt | llm | parser

res = chain.invoke({"topic": "프롬프트 엔지니어링"})

print(res)
```
