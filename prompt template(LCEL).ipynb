```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage

# env 파일에 저장된 환경번수 로드
load_dotenv()

ai_key = os.getenv("OPENROUTER_API_KEY")

llm = ChatOpenAI(
    model='openrouter/free',   # select free model
    api_key=ai_key,
    base_url="https://openrouter.ai/api/v1"
)

msg = [
    SystemMessage(content="너는 Python 강사이고, 2문장 이상 말하지 않아"),
    HumanMessage(content="Python이 무엇인가요?")
]

res = llm.invoke(msg)

print(res.content)
```

```python
# prompt template / LCEL

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# 각 처리 단계를 파이프 기호로 연결

prompt = ChatPromptTemplate([
    ("system", "너는 핵심만 설명하는 AI 강사야. 최대한 자세히 설명하되 3문장을 넘지 않아"),
    ("human", "{topic}을 초보자 수준으로 설명해줘")
])

parser = StrOutputParser()

chain = prompt | llm | parser

res = chain.invoke({"topic":"생성형 AI"})

print(res)
```
